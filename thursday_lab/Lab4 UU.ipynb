```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from scipy.special import gamma
import ipywidgets as widgets

rng = np.random.default_rng(seed=100)
```

#Q1

```python
K = 5000
n = 5000
grid = np.arange(1, K+1) / K
grid

x = np.sort(rng.choice(grid, size = n))
x
plt.step(x, np.arange(1, n+1)/n) 

x = np.linspace(0, 1, 200)
plt.plot(x, stats.uniform.cdf(x, loc=0, scale=1-0))
```

as K increases it resembles a uniform distribution

#Q2

```python
n = 32
b = 5000

z = []
for i in range(b):
    rand = rng.uniform(-np.sqrt(3), np.sqrt(3), size=n)
    randmean = rand.mean()
    p3 = rand.mean() * np.sqrt(n)
    z.append(p3)
x = np.sort(z)
plt.step(x, np.arange(1, b+1)/b)


x = np.linspace(-4, 4, 200)
plt.plot(x, stats.norm.cdf(x, loc=0, scale=1))


```

as n increases it resembles a normal distribution

#Q3

```python
n = 40
p = .5
b = 5_000

z = []
for i in range(b):
    flip = rng.choice([0,1], size = n)
    heads = flip.sum()
    pheads = n * p
    sd = np.sqrt(n * p * (1-p))
    z.append((heads - pheads)/sd)

x = np.sort(z)
plt.step(x, np.arange(1, 1+b)/b)

x = np.linspace(-4, 4, 200)
plt.plot(x, stats.norm.cdf(x, loc=0, scale=1))

```

as n increases it resembles a normal distribution

as p increases .5 -> 1 the distribution skews negative
as p decreases .5 -> 0 the distribution skews postive

```python
dt = 1e-3
n = 5_000
r = 1
p = r * dt

periods = rng.geometric(p, size = n)
times = periods * dt

x = np.sort(times)
plt.step(x, np.arange(1, n+1)/n)

x = np.linspace(0, 8, 200)
plt.plot(x, stats.expon.cdf(x, scale=1/r))
```

as dt goes to zero it resembles a exponetial distribution

as r gets closer to zero the x axis becomes larger
